```python
import torch
from diffusers import StableDiffusionImg2ImgPipeline

model_id = "runwayml/stable-diffusion-v1-5"
try:
    # Load the pipeline with the required configuration
    pipe = StableDiffusionImg2ImgPipeline.from_pretrained(model_id, torch_dtype=torch.float16).to(device)
    print("Model loaded successfully!")
except Exception as e:
    print(f"Error loading model: {e}")
```

```python
def process_image_edit(image_path, prompt, strength=0.75, guidance_scale=7.5):
    # Load and prepare the image
    init_image = Image.open(image_path).convert("RGB")
    init_image = init_image.resize((512, 512))

    # Run the image-to-image pipeline
    with torch.autocast(device):
        result = pipe(prompt=prompt, image=init_image, strength=strength, guidance_scale=guidance_scale).images[0]

    return result
```

```python
from google.colab import files
import matplotlib.pyplot as plt
import os

print("لطفاً یک تصویر (jpg, png) برای ویرایش انتخاب کنید:")
uploaded = files.upload()

for fn in uploaded.keys():
    if fn.lower().endswith(('.png', '.jpg', '.jpeg', '.webp')):
        print(f'در حال پردازش تصویر: "{fn}"')
        prompt = "a person wearing a futuristic silver space suit, highly detailed, cinematic lighting, 8k"
        try:
            result_image = process_image_edit(fn, prompt, strength=0.6)

            # نمایش نتیجه
            plt.figure(figsize=(10, 5))
            plt.imshow(result_image)
            plt.axis('off')
            plt.title("نتیجه ویرایش هوشمند")
            plt.show()

            output_name = f"edited_{fn}"
            result_image.save(output_name)
            print(f"فایل با نام {output_name} ذخیره شد.")
        except Exception as e:
            print(f"خطا در پردازش تصویر: {e}")
    else:
        print(f"فایل {fn} یک تصویر معتبر نیست. لطفاً فایل تصویر انتخاب کنید.")
```
